```python
# Load the dataset
import pandas as pd

# Read male players data
df = pd.read_csv("../data/raw/Dataset/male_players.csv", low_memory=False)

print(f"Shape: {df.shape}")
print(f"\nFirst 3 rows:")
df.head(3)
```

```python
print(df['fifa_version'].unique())
```

```python
# Filter FIFA 24 only
df = df[df['fifa_version'] == 24]

# Select only relevant columns
cols = [
    # Identity
    'short_name', 'long_name', 'age', 'nationality_name', 'player_positions',
    # Club & League
    'club_name', 'league_name', 'league_level',
    # Market Value
    'value_eur', 'wage_eur', 'release_clause_eur',
    # Overall
    'overall', 'potential', 'international_reputation',
    # Key Skills
    'pace', 'shooting', 'passing', 'dribbling', 'defending', 'physic',
    # Detailed Skills
    'preferred_foot', 'weak_foot', 'skill_moves',
    'attacking_finishing', 'attacking_short_passing', 'attacking_crossing',
    'skill_dribbling', 'skill_ball_control', 'mentality_vision',
    'mentality_composure', 'power_stamina', 'power_strength',
    'movement_acceleration', 'movement_sprint_speed', 'movement_agility',
    'defending_marking_awareness', 'defending_standing_tackle',
    'mentality_positioning', 'work_rate', 'body_type'
]

df = df[cols]
df = df.reset_index(drop=True)

print(f"Shape after filtering: {df.shape}")
df.head(3)
```

```python
# Save filtered data to raw folder
df.to_csv(
    "../data/raw/fifa24_players.csv",
    index=False
)

print("File saved successfully!")
print(f"Total Players: {len(df)}")
print(f"Total Columns: {len(df.columns)}")
print(f"\nLeagues in dataset:")
print(df['league_name'].value_counts().head(10))
```

```python
# Check our target leagues
target_leagues = ['Premier League', 'La Liga', 'Bundesliga', 'Serie A', 'Ligue 1']

for league in target_leagues:
    count = len(df[df['league_name'] == league])
    print(f"{league}: {count} players")
```

```python
# Filter only Top 5 European Leagues
top5_leagues = ['Premier League', 'La Liga', 'Bundesliga', 'Serie A', 'Ligue 1']

df_top5 = df[df['league_name'].isin(top5_leagues)]
df_top5 = df_top5.reset_index(drop=True)

print(f"Total players in Top 5 leagues: {len(df_top5)}")
print(f"\nTop 10 players by overall rating:")
print(df_top5[['short_name', 'club_name', 'league_name', 'overall', 'value_eur']]
      .sort_values('overall', ascending=False)
      .head(10)
      .to_string())
```

```python
# Save Top 5 leagues dataset
df_top5.to_csv(
    "../data/cleaned/fifa24_top5_leagues.csv",
    index=False
)

print("✅ Main dataset saved!")
print(f"Players: {len(df_top5)}")
print(f"Leagues: {df_top5['league_name'].nunique()}")
print(f"Clubs: {df_top5['club_name'].nunique()}")
print(f"Nationalities: {df_top5['nationality_name'].nunique()}")
```
